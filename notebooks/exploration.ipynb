# SZR Outbreak Predictor — Exploration

This notebook provides exploratory data analysis for the synthetic SZR dataset
generated by `data/generate_data.py`.

**Run order before opening this notebook:**
1. `python data/generate_data.py`
2. `python model/train.py`

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style='whitegrid')
```

## Load Dataset

```python
DATA_PATH = os.path.join('..', 'data', 'szr_synthetic.csv')
df = pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
```

## Basic Statistics

```python
df.describe()
```

## Class Balance (Containment)

```python
counts = df['containment'].value_counts()
print(counts)
counts.plot(kind='bar', title='Containment class balance', rot=0)
plt.ylabel('Count')
plt.show()
```

## Target Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['peak_zombie_fraction'].hist(bins=50, ax=axes[0])
axes[0].set_title('Peak Zombie Fraction')
df['time_to_peak'].hist(bins=50, ax=axes[1])
axes[1].set_title('Time to Peak (days)')
plt.tight_layout()
plt.show()
```

## Feature Correlations

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature & Target Correlations')
plt.tight_layout()
plt.show()
```
